```python
zero_quantity = data[data["Quantity"] == 0]

zero_quantity.shape
```

```python
zero_price = data[data["UnitPrice"] == 0]

zero_price.shape
```

```python
zero_price.head(20)
```

```python
clean_data = data.copy()
```

```python
# find zero price transactions with customer ID not null

# read data and look at CustomerID, if the row has a CustomerID, filter it out.
zero_price_with_customer = zero_price[zero_price["CustomerID"].notnull()]

zero_price_with_customer.shape
zero_price_with_customer.head(20)
```

```python
# filter out rows where Customer ID is not missing
clean_data = clean_data[clean_data["CustomerID"].notnull()]

# Remove cancelled txns
clean_data = clean_data[clean_data["Quantity"] > 0]

# Remove zero or negative prices
clean_data = clean_data[clean_data["UnitPrice"] > 0]

clean_data.shape
```

```python
clean_data.isnull().sum()
```

```python
clean_data.describe()
```

```python
clean_data.duplicated().sum()
```

This means there are 5192 rows that are exact duplicates (of an earlier row).

```python
duplicate_rows = clean_data[clean_data.duplicated()]

duplicate_rows.shape
```

```python
duplicate_rows.head(20)
```

There are 5192 exact duplicate rows. For the purpose of this project, given that there's no additional information to determine whether these rows represent separate purchases or duplicate records, we will remove and exclude these rows to avoid potentially overstating customers' purchase frequency and spending value. 

```python
# remove exact duplicate rows
clean_data = clean_data.drop_duplicates()

# check that duplicates have been removed
clean_data.duplicated().sum()
```
