```python
import pandas as pd
```

```python
data = pd.read_csv("clean_data.csv")
data.head()
```

```python
# count the number of unique customers in dataset

data["CustomerID"].nunique()
```

```python
data["Revenue"] = data["Quantity"] * data["UnitPrice"]

data.head()
```

```python
# What is the total revenue?

data["Revenue"].sum()
```

```python
total_revenue = data["Revenue"].sum()

print(round(total_revenue, 2))
```

The cleaned dataset contains 4338 unique customers and generated approximately 8.89M pounds during the period analysed.

```python
# group the data by country and calc total revenue for each country

revenue_by_country = data.groupby("Country")["Revenue"].sum()

# sort
revenue_by_country = revenue_by_country.sort_values(ascending=False)

revenue_by_country
```

```python
uk_revenue = revenue_by_country["United Kingdom"]

uk_percentage = uk_revenue / total_revenue * 100

print(round(uk_percentage, 2))
```

Analysis: Revenue is heavily concentrated in the UK. International markets collectively contribute a much smaller share of revenue.

```python
international_revenue = revenue_by_country[revenue_by_country.index != "United Kingdom"]

international_revenue.head(5)
```

```python
# convert data type of InvoiceDate from string to datetime

data["InvoiceDate"] = pd.to_datetime(data["InvoiceDate"])
data["InvoiceDate"].dtype
data["InvoiceDate"].head()
```

```python
# create new column for Month

data["Month"] = data["InvoiceDate"].dt.to_period("M")

data.head()
```

```python
# group data by month and calc total revenue for each month

revenue_by_month = data.groupby("Month")["Revenue"].sum()

revenue_by_month
```

Note: Earlier in the other notebook -> Earliest dated txn: Dec 1 2010, Latest dated txn: Dec 9 2021
So 517190.440 represents the revenue for Dec 2021 as of 9 Dec (9 day revenue).

Analysis: Monthly revenue remained relatively stable through early and mid-2011 before increasing substantially from September onwards. Revenue reached approx. £1.16M in November 2011, the highest full month in the dataset and roughly 80% higher than August 2011. 

This suggests higher purchasing activity toward the year-end period in Q4 potentially due to holiday/gifting season. December 2011 was excluded from month-to-month interpretation because the dataset only contains transactions through 9 December.

```python
# calculate number of unique orders for each month

orders_by_month = data.groupby("Month")["InvoiceNo"].nunique()

orders_by_month
```

It seems the year-end surge in revenue was substantially driven by an incease in number of orders.

```python
aug_revenue = revenue_by_month["2011-08"]
nov_revenue = revenue_by_month["2011-11"]

revenue_uplift = (nov_revenue - aug_revenue) / aug_revenue * 100

print("Revenue uplift:", round(revenue_uplift, 2), "%")

aug_orders = orders_by_month["2011-08"]
nov_orders = orders_by_month["2011-11"]

order_uplift = (nov_orders - aug_orders) / aug_orders * 100

print("Order uplift:", round(order_uplift, 2), "%")
```

Orders grew faster than revenue. So the number of orders increased by the ABS fell during this period (Aug - Nov).

```python
# calculate average order value for each month

average_order_value = revenue_by_month / orders_by_month

average_order_value
```

```python
aug_abs = average_order_value["2011-08"]
nov_abs = average_order_value["2011-11"]

abs_change = (nov_abs - aug_abs) / aug_abs * 100

print("August ABS: £", round(aug_abs, 2))
print("November ABS: £", round(nov_abs, 2))
print("ABS change:", round(abs_change, 2), "%")
```

Revenue growth from August to November was primarily driven by increased order volume rather than higher spending per order. While we observe an order uplift of 108% (no. of orders more than doubled), ABS declined by about 14% over the same period.

Conclusions:

Customer base: 4338

Total revenue: £8.89M

UK share of revenue: 82%

Trends: Strong revenue observed in Q4 months, driven by higher order volume

```python
# combine recency, frequency and monetary into one table
rfm = pd.concat([recency, frequency, monetary], axis=1)

# rename the columns because we prev grouped them by 
# the invoice date (R), number of unique orders (F) and revenues (M)
rfm.columns = ["Recency", "Frequency", "Monetary"]

rfm.head(10)
```

```python
rfm.tail(10)
```

### RFM SCORING

```python
# R scoring: divide customers into 5 groups based on recency
# 5 = most recent customers, 1 = least recent customers

rfm["R_Score"] = pd.qcut(
    rfm["Recency"],
    5,
    labels=[5, 4, 3, 2, 1]
)

rfm.head(10)
```

```python
# rfm["F_Score"] = pd.qcut(
    # rfm["Frequency"],
    # 5,
    # labels=[1, 2, 3, 4, 5]
# )

# code above failed because there are many customers with the same frequency value, so the qtys cannot be calculated properly. 
# Instead, we can use value_counts to see the distribution of frequency values.

rfm["Frequency"].value_counts().sort_index()
```

Analysis: Frequency data is very uneven. Out of 4338 customers, 1493 only made one unique order (34%). This makes it difficult to divide customers into equal-sized groups, so I will use custom frequency bins instead based on purchase behaviour.

```python
def frequency_score(frequency):

    if frequency == 1:
        return 1

    elif frequency == 2:
        return 2

    elif frequency <= 4:
        return 3

    elif frequency <= 9:
        return 4

    else:
        return 5

rfm["F_Score"] = rfm["Frequency"].apply(frequency_score)

rfm.head(10)
```

```python
# M scoring
# 5 = highest spending customers, 1 = lowest spending customers

rfm["M_Score"] = pd.qcut(
    rfm["Monetary"],
    5,
    labels=[1, 2, 3, 4, 5]
)

rfm.head(10)
```

```python
# check how many customers received each score

print("R Scores:")
print(rfm["R_Score"].value_counts().sort_index())

print("\nF Scores:")
print(rfm["F_Score"].value_counts().sort_index())

print("\nM Scores:")
print(rfm["M_Score"].value_counts().sort_index())
```

Analysis: Customers are relatively evenly distributed across the Recency and Monetary scores. Frequency is more uneven because a large proportion of customers made only one or a few purchases, so custom bins were used to better reflect actual purchasing behaviour.

## CUSTOMER SEGMENTATION

```python
# calculate total RFM score for each customer
# convert to type int

rfm["RFM_Score"] = (
    rfm["R_Score"].astype(int)
    + rfm["F_Score"].astype(int)
    + rfm["M_Score"].astype(int)
)

rfm.head(10)
```

```python
# create customer segments based on RFM scores

def customer_segment(row):

    # if customer purchsed recently, purchases frequently and spends a lot
    if row["R_Score"] >= 4 and row["F_Score"] >= 4 and row["M_Score"] >= 4:
        return "High-Value Loyal"

    # if customer purchased recently, but hasn't purchased frequently
    elif row["R_Score"] >= 4 and row["F_Score"] <= 2:
        return "Recent / Low Frequency"

    # if customer hasn't purchased recently, but historically purchased frequently
    elif row["R_Score"] <= 2 and row["F_Score"] >= 4:
        return "Risk of Churn"

    # if customer hasn't purchased recently and hasn't purchased frequently
    elif row["R_Score"] <= 2 and row["F_Score"] <= 2:
        return "Inactive / Low Frequency"
    
    else:
        return "Regular"
```

```python
rfm["Segment"] = rfm.apply(customer_segment, axis=1)

rfm.head(10)
```

```python
# count number of customers in each segment

segment_counts = rfm["Segment"].value_counts()

segment_counts
```

```python
# calculate percentage of customers in each segment

segment_percentage = segment_counts / len(rfm) * 100

segment_percentage.round(2)
```

Analysis: Regular customers form the largest segment at 36.72% of the customer base. However, 30.77% of customers are inactive and have historically purchased infrequently, suggesting a sizeable portion of the customer base has relatively low engagement. 

Meanwhile, 11.87% are recent but low-frequency customers, presenting an opportunity to encourage repeat purchases. High-Value Loyal customers account for 18.19% and represent an important segment to retain and reward. Only 2.44% are classified as Risk of Churn, representing previously frequent customers who have not purchased recently.

```python
# calculate total revenue generated by each customer segment

revenue_by_segment = rfm.groupby("Segment")["Monetary"].sum()

# sort from highest to lowest revenue

revenue_by_segment = revenue_by_segment.sort_values(ascending=False)

revenue_by_segment
```

High-Value Loyal customers generate about £5.44M, even though they are only 18.19% of customers. Meanwhile, Inactive / Low Frequency customers are 30.77% of customers, but generate only about £682K.

```python
# calculate total revenue across all segments

total_segment_revenue = revenue_by_segment.sum()

# calculate percentage of revenue generated by each segment

revenue_percentage = revenue_by_segment / total_segment_revenue * 100

revenue_percentage.round(2)
```

Analysis: Revenue is highly concentrated among High-Value Loyal customers, who account for only 18.19% of the customer base but generate 61.26% of total revenue. 

In contrast, Inactive / Low Frequency customers represent 30.77% of customers but contribute only 7.67% of revenue. 

CRM strategies should prioritise retaining high-value loyal customers, while reactivation efforts for inactive customers may need to be more targeted. There also seems to be some opportunity to turn some regular customers into high-value, loyal customers. Based on prior findings, regular users make up 36.72% of total customers and generate 23.55% of revenue.

```python
# turn all data into table for easier analysis

# create an empty table
segment_summary = pd.DataFrame()

# add number of customers
segment_summary["Customers"] = segment_counts

# add percentage of customers
segment_summary["% Customers"] = segment_percentage.round(2)

# add total revenue
segment_summary["Revenue"] = revenue_by_segment.round(2)

# add percentage of revenue
segment_summary["% Revenue"] = revenue_percentage.round(2)

segment_summary
```

It is important to concentrate strategies that target retaining high-value loyal customers (highly valuable customer segment). We should also identify high-value loyal customers at risk of churn.

```python
# filter for customers who are at risk of churn

risk_customers = rfm[rfm["Segment"] == "Risk of Churn"]

risk_customers.head(10)
```

```python
# sort at-risk customers from highest to lowest spending

risk_customers = risk_customers.sort_values(
    "Monetary",
    ascending=False
)

risk_customers.head(10)
```

```python
# calculate revenue from top 10 highest-spending at-risk customers

top_10_risk_revenue = risk_customers.head(10)["Monetary"].sum()

print("Top 10 at-risk customer revenue: £", round(top_10_risk_revenue, 2))
```

```python
# calculate total revenue from all at-risk customers

total_risk_revenue = risk_customers["Monetary"].sum()

# calculate percentage contributed by top 10

top_10_percentage = top_10_risk_revenue / total_risk_revenue * 100

print("Top 10 share of at-risk revenue:", round(top_10_percentage, 2), "%")
```

Analysis: The top 10 highest-spending at-risk customers account for 27.58% of total revenue generated by the Risk of Churn segment. This suggests that the value of at-risk customers is relatively concentrated among a small group. 

Reactivation efforts could and should prioritise these historically high-value customers.

## MARKETING RECOMMENDATIONS

Based on the RFM customer segmentation, different CRM strategies can be used for each customer segment to improve retention, reactivation and customer value.

### High-Value Loyal
#### Strategy: Retain & Reward

High-Value Loyal customers represent only 18.19% of customers but contribute 61.26% of total revenue, making them the most commercially valuable customer segment.

**Recommended actions:**
- Prioritise retention through loyalty rewards and exclusive benefits.
- Provide early access to new products or promotions.
- Personalise product recommendations based on previous purchases.
- Avoid relying excessively on large discounts, as these customers already demonstrate high purchase frequency and spending. There is no need to focus budget on discounts for this segment.

### Risk of Churn
#### Strategy: Winback

Risk of Churn customers:
106 customers | 2.44% of customers | 2.77% of revenue

Although RoC customers represent only 2.44% of the customer base, these customers have historically purchased frequently but have not purchased recently.

The top 10 highest-spending customers alone account for 27.58% of the segment's historical revenue, suggesting that reactivation efforts should be prioritised based on customer value.

**Recommended actions:**
- Target high-value at-risk customers with personalised win-back campaigns.
    - E.g. Segment the high-value at-risk customers into different categories based on purchase history (Fashion consumers, Lifestyle, Beauty, etc.), explore specific category vouchers
- Offer stronger incentives selectively to historically valuable customers.
- Use lower-cost CRM channels such as email reminders for lower-value at-risk customers.
- Monitor whether reactivated customers make another purchase after receiving the campaign.
- Can potentially explore A/B testing: Targeted incentives/messaging vs. standard win-back message.

### Recent / Low Frequency
#### Strategy: Encourage Repeat Purchase

Recent / Low Frequency customers represent 11.87% of customers and contribute 4.74% of total revenue. These customers have purchased recently but have relatively low purchase frequency, presenting an opportunity to encourage a second or subsequent purchase.

**Recommended actions:**
- Post-purchase comms encouraging repeat purchases.
- Provide personalised product recommendations based on their recent purchase.
- Use time-limited return vouchers or second-purchase incentives.
- Introduce customers to loyalty benefits to encourage longer-term engagement.
- Introduce cashback/return vouchers (note not to utilise too much marketing budget on this as this customer segment is not as valuable).

### Inactive / Low Frequency
#### Strategy: Low-Cost Reactivation (only if budget allows)

Inactive / Low Frequency customers make up 30.77% of the customer base but contribute only 7.67% of total revenue. Given their relatively low historical value and lack of recent activity, costly reactivation incentives may not be justified for the entire segment.

**Recommended actions:**
- Use low-cost CRM channels such as email for reactivation campaigns.
- Promote relevant new products or seasonal campaigns to encourage customers to return.
- Test small incentives before offering larger discounts.
- Prioritise marketing spend towards customers who demonstrate renewed engagement.

### Regular
#### Strategy: Grow Customer Value

Regular customers form the largest segment at 36.72% of customers and contribute 23.55% of total revenue. These customers demonstrate moderate purchasing behaviour and represent an opportunity to grow future customer value.

**Recommended actions:**
- Use personalised recommendations and cross-selling to increase basket size.
- Encourage more frequent purchases through targeted CRM campaigns.
- Promote loyalty programme benefits or spending milestones.
- Identify customers whose behaviour is approaching High-Value Loyal status and target them with relevant incentives.
